```python
from langchain_community.document_loaders import PyPDFLoader
loader = PyPDFLoader("AdaBM On-the-Fly Adaptive Bit Mapping for Image Super-Resolution.pdf")

docs = loader.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=20)
text_splitter.split_documents(docs)[:5]
```

```python
documents = text_splitter.split_documents(docs)
documents
```

```python
from langchain_openai.embeddings import OpenAIEmbeddings
from langchain_ollama.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

# db = FAISS.from_documents(documents, OpenAIEmbeddings())
db = FAISS.from_documents(documents, OllamaEmbeddings(model="qwen3-embedding:8b"))
```

```python
db
```

```python
query="what is the topic"
result = db.similarity_search(query)
result[:5]
```

```python
from langchain_ollama import ChatOllama

llm = ChatOllama(model="qwen3.8")
llm
```

```python
### Designing chat prompt template

from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_template("""
Answer the following question based only on the provided context,
Think step by step before providing a detailed answer.
I will tip you $1000 if the user find the answer helpful.
<context>
{context}
</context>
Question: {input}
""")
```

```python
## chain introduction
## Create stuff document  

from langchain_classic.chains.combine_documents import create_stuff_documents_chain

document_chain = create_stuff_documents_chain(llm, prompt)
```

```python
retriever = db.as_retriever()
retriever
```

```python
## retriever chain

from langchain_classic.chains import create_retrieval_chain

retriever_chain = create_retrieval_chain(retriever, document_chain)
```

```python
response = retriever_chain.invoke({"input": "What is architectur of this article?"})
```
