```python
## Data ingestion

from langchain_community.document_loaders.text import TextLoader

loader = TextLoader("legendary_prehistory_characters.txt")
text_document = loader.load()
text_document
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
## langsmith tracking
os.environ["LANGCHAIN_TRACING_V2"] = "true"
os.environ["LANGCHAIN_API_KEY"] = os.getenv("LANGCHAIN_API_KEY")
```

```python
## Web based loader
from langchain_community.document_loaders import WebBaseLoader
import bs4

## load ,chunck and index the content of the html page
loader = WebBaseLoader(web_paths=("https://lilianweng.github.io/posts/2023-06-23-agent/",),
                       bs_kwargs=dict(parse_only=bs4.SoupStrainer(
                           class_=("post-title", "post-content", "post-header")
                       )),)

text_document = loader.load()
text_document
```

```python
from langchain_community.document_loaders import PyPDFLoader

loader = PyPDFLoader("AdaBM On-the-Fly Adaptive Bit Mapping for Image Super-Resolution.pdf")

text_document = loader.load()
text_document
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_spilliter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
docs = text_spilliter.split_documents(text_document)
docs[-5:]
```

```python
docs
```

```python
## Vextor Embedding and Vector Store
from langchain_openai import OpenAIEmbeddings
from langchain_ollama import OllamaEmbeddings
from langchain_chroma import Chroma

# db = Chroma.from_documents(docs[:20], OpenAIEmbeddings())
db = Chroma.from_documents(docs[:20], OllamaEmbeddings(model="qwen3-embedding:4b"))
```

```python
## Vector database

query="Who are the authors of article"
result = db.similarity_search(query) 
result[0].page_content
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_community.vectorstores import FAISS

db = FAISS.from_documents(docs[:20], OllamaEmbeddings(model="qwen3-embedding:4b"))
```

```python
## Vector database

query="Who are the authors of article"
result = db.similarity_search(query) 
result[0].page_content
```
